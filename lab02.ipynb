Bài 3: Kiểm tra thông tin Tensor

torch.size([2,3,4]) có nghĩa là 2 ma trận trong đó mỗi ma trận có 3 hàng và 4 cột. và phương thức Size() có nghĩa là cung cấp kích thước trên từng chiều. Giúp biết được cấu trúc tổ chức dữ liệu bên trong Tensor đó như thế nào trước khi thực hiện các pháp toán

```python
import torch
a = torch.zeros(2, 3, 4)
print("Tensor:")
print(a)
print("Shape:", a.shape)
print("Size:", a.size())
print("Number of dimensions:", a.ndim)
print("Data type:", a.dtype)
```

Bai 4: Slicing va Indexing 

a) với a[1] thực hiện chỉ mục tại chiều đầu tiên nghĩa là chọn ra ma trận thứ 2 với index = 1 nên kết quả thu được là ma trận Thứ 2 trong tổng số 3 ma trận và có 4 hàng và 5 cột
- b) với a[1:, 2:4] thì đây là phép cắt lát trên từng chiều
- Với chiều đầu tiên (chiều 0) là (1:) thì lấy từ chỉ số 1 đến hết nghĩa là lấy ma trận thứ 2,3 bỏ ma trận 1 là chỉ số 0
- Chiều 1 (2:4) là chỉ lấy hàng 2 và trước 4 (gồm hàng 2 và 3) 
- Chiều 2 (kh viết): Mặc định giữ nguyên toàn bộ là đầy đủ 5 cột


```python
import torch
a = torch.randn(3, 4, 5)
print("Original tensor:")
print(a)
print("\nTensor a[1]:")
print(a[1])
print("\nTensor a[1:, 2:4]:")
print(a[1:, 2:4])
print("Hai phan tu cuoi tren chieu thu ba:")
print(a[:, :, -2:])
```

Bài 5: Resphae và Flatten

```python
from torch._inductor import cudagraph_trees
import torch
a = torch.randn(3, 4)
print("Original:")
print(a)
print("\nFlatten:")
print(a.ravel())
a_reshaped = a.reshape(3, 2, 2)
print("\nReshape to 3 x 2 x 2:")
print(a_reshaped)

#Kiem tra so phan tu
print("So phan tu truoc khi reshape:", a.numel())
print("So phan tu sau khi reshape:", a_reshaped.numel())
#2. Thu reshape Tensor thanh kich thuoc 2x6
a_2x6 = a.reshape(2, 6)
print("Reshape thanh 2 x6:")
print(a_2x6)
```
