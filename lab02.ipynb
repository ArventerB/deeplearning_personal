Bài 7: Các hàm thống kê trên Tensor
- dim = 0
    - Tính theo chiều dọc theo các cột, đi qua các hàng và gồm 4 giá trị đại diện cho cả 4 cột và sizee lúc này sẽ được cộng dồn từ hàng trên xuống hàng dưới
    - Bên cạnh đó do tensor có 4 cột nên kết quả trả về là một vector gồm 4 phần tử
- dim = 1
    - Tính theo chiều ngang theo các hàng, đi qua các cột và gồm 3 giá trị đại diện cho 3 hàng và size sẽ được cộng dồn từ cột trái sang cột phải
    - Bên cạnh đó do có 3 hàng, kết quả trả về một vector gồm 3 phần tử

```python
import torch
a = torch.randn(3, 4)
print(a)
print("\nMean:")
print(torch.mean(a, dim=0))
print("\nStandard deviation:")
print(torch.std(a, dim=0))
print("\nCumulative sum:")
print(torch.cumsum(a, dim=0))

print("\nMean dim = 1:")
print(torch.mean(a, dim=1))
print("\nStandard deviation dim = 1:")
print(torch.std(a, dim=1))
print("\nCumulative sum dim = 1:")
print(torch.cumsum(a, dim=1))
```

Bài 8: Autograd cơ bản
- Tính thủ công đạo hàm của y = x2 tại x = 3.6 và so sánh với x.grad.
    - Đạo hàm của y =x^2 => 2x và tại x = 3.6 nên y = 7.2
    - Với x.grad cũng = 7.2 nên ta thấy 2 kết quả giống nhau
- Đổi hàm thành y = 3*x**2 + 2*x + 1 và kiểm tra gradient.
    - Giá trị gradient được tính ra x.grad = 23.6 và tính thủ công cũng = với tính tay

```python
import torch
x = torch.tensor(
3.6,
requires_grad=True
)
y = 3*x**2 + 2*x + 1
y.backward()
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
```

Bài 9. Gradient Descent tìm hệ số đa thức

```python
import numpy as np
import torch
polynomial = np.poly1d([1, 2, 3])
N = 20

X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
print("Learned coefficients:")
print(w)
```
