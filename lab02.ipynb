- dim = 0
    - Tính theo chiều dọc theo các cột, đi qua các hàng và gồm 4 giá trị đại diện cho cả 4 cột và sizee lúc này sẽ được cộng dồn từ hàng trên xuống hàng dưới

Bài 9. Gradient Descent tìm hệ số đa thức
- a. So sánh ba hệ số học được với [1,2,3]
    - Trước khi huấn luyện: Trọng số w được khởi tạo ngẫu nhiên nên có giá trị sai lệch lớn so với bộ hệ số [1,2,3] ban đầu
    - Sau 1000 epochs huấn luyện: Hệ số của x^2 hội tụ nhanh và sát nhất (đạt 0.9363 gần với 1) vì bậc 2 có độ biến thiên rớt nhất nên ảnh hưởng mạnh nhất đến hàm mất mát MSE
        - Hệ số x và hệ số tự do cũng đều tự điều chỉnh và tiến về gần về phía mục tiêu (lần lượt tiệm cận về mốc 2 và 3)
    - Từ đó ta có thể thấy Thuật toán tối ưu Gradient Descent đã hoạt động hiệu quả giúp mô hình tự học và tìm lại gần đúng hàm đa thức gốc ban đầu từ dữ liệu mẫu. Nếu tăng lần lặp (epoch) lên 2000 - 3000 thì bộ hệ số này sẽ hội tụ chính xác tuyệt đối về [1.0, 2.0, 3.0]

- b. Giải thích vai trò của optimizer.zero_grad(), mse.backward() và optimizer.step().
    - optimizer.zero_grad(): có vai trò đặt lại gradient của tất cả các tham số cần học về giá trị 0 trước khi bắt đầu tính toán cho vòng lặp mới. Nó bắt buộc phải có vì cơ chế mặc định là cộng dồn gradient qua mỗi lần gọi .backward(). Nếu không xóa về 0 ở đầu mỗi epoch, gradient của vòng lặp hiện tại sẽ bị cộng dè lên gradient của các vòng lặp trước đó nên sẽ khiến các bước nhảy bị sai lệch hoàn toàn và mô hình không thể hội tụ. Kết quả tính được sẽ được lưu trữ tự động vào thuộc tính w.grad.
    - mse.backward(): là lan truyền ngược có vai trò thực hiện quá trình lan truyền ngược để tính toán cho hàm (gradient) của hàm mất mát MSE theo từng hệ số trong w,
    - optimizer.step(): là Cập nhật trọng số và có vai trò thực hiện thay đổi, cập nhật giá trị của các hệ số w nhằm giảm thiểu sai số. Nên nhờ lệnh này mà sau mỗi epoch, các giá trị trong w ngày càng nhích gần hơn về bộ số [1,2,3].

```python
import numpy as np
import torch
polynomial = np.poly1d([1, 2, 3])
N = 20

X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
print("Learned coefficients:")
print(w)
```

Bài 10: Giải hệ phương trình bằng Autograd
a) Kiểm tra nghiệm thu được với bốn phương trình: A+B=9, C-D=1, A+C=8, B-D=2
   - A gần bằng 4.4434
    - B gần bằng 4.5565
    - C gần bằng 3.5566
    - D gần bằng 2.5565
- Vậy khi thay các giá trị trên vào 4 phương trình theo đề cho đều thỏa mãn

b) Nhận xét vì sao bài toán có nhiều nghiệm:
- Ta thấy có 3 phương trình độc lập tuyến tính nhưng có tới 4 ẩn số là A B C D nên ta có số ẩn nhiều hơn số phương trình (4 > 3) nên có 1 bậc tự do và sẽ tôn tại vô số nghiệm
- Ngoài ra do bài toán có vô số nghiệm, đáy của hàm mất mát không phải là một điểm đơn lẻ mà là cả một đường thẳng liên tục trong không gian 4 chiều. Các biến A B C D ban đầu được gán vào giá trị ngẫu nhiên. Tùy thuộc vào thời điểm khởi tạo ban đầu, thuật toán Gradient Des sẽ trượt xuống và hội tụ tạo một điểm bất kỳ trên đường thẳng của nghiệm đó. Vì vậy mỗi lần chạy lại chương trình thì ta lại nhận một bộ nghiệm số khác nhau nhưng tất cả đều đúng



```python
from numpy._typing import _nested_sequence
import random
import torch
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)
for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()
print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)

print("Nghiem thu duoc voi bon phuong trinh:")
print("PT1: (A + B = 9):", A.item() + B.item())
print("PT 2 (C - D = 1):", C.item() - D.item())
print("PT 3 (A + C = 8):", A.item() + C.item())
print("PT 4 (B - D = 2):", B.item() - D.item())
print("Tong sai so:", sqerr.item())
```

```python
import numpy as np  
import torch    

print("YC 1: Tạo tensor X có 24 phần tử và reshape lần lượt thành 4 x6, 2 x 12 và 2 x3 x4")

x = torch.arange(24)
print("Tensor x ban dau: ", x)
print("Shape ban dau: ", x.shape)

x_4x6 = x.reshape(4, 6)
print("Reshape 4 x 6: \n", x_4x6)

x_2x12 = x.reshape(2, 12)
print("Reshape 2 x 12: \n", x_2x12)

x_2x3x4 = x.reshape(2, 3, 4)
print("Reshape 2 x 3 x 4: \n", x_2x3x4)

print("YC2: CÁC PHÉP TOÁN TRÊN TENSOR A VÀ B (3 x 4)")
torch.manual_seed(42)
A = torch.randint(1, 10, (3, 4)).float()
B = torch.randint(1, 10, (3, 4)).float()

print("Tensor A:\n", A)
print("Tensor B:\n", B)
print("\nPhép cộng A + B:\n", A + B)
print("\nNhân từng phần tử A * B:\n", A * B)
print("\nChia từng phần tử A / B:\n", A / B)
# Trung bình theo cột
mean_col_A = torch.mean(A, dim=0)
print("\n4. Trung bình theo cột của A (dim=0):\n", mean_col_A)
# Độ lệch chuẩn theo cột
std_col_A = torch.std(A, dim=0)
print("\n5. Độ lệch chuẩn theo cột của A (dim=0):\n", std_col_A)

print("YC3: TÍNH ĐẠO HÀM y = 2x^2 + 5x + 3 TẠI x = 2")
x = torch.tensor(2.0, requires_grad=True)
y = 2 * x**2 + 5 * x + 3
y.backward()
print(f"Giá trị tính bằng Autograd (x.grad): {x.grad.item()}")

# Tính thủ công:
# y' = d(2x^2 + 5x + 3)/dx = 4x + 5
# Tại x = 2: y'(2) = 4*(2) + 5 = 13
manual_grad = 4 * 2 + 5
print(f"Giá trị tính thủ công: 4*2 + 5 = {manual_grad}")
print(
    f"So sánh: {'Trùng khớp hoàn toàn' if x.grad.item() == manual_grad else 'Khác nhau'}"
)

print("YC 4: TÌM HỆ SỐ ĐA THỨC y = 2x^2 - 3x + 5")
np.random.seed(42)
torch.manual_seed(42)
polynomial = np.poly1d([2, -3, 5]) #đa thức theo đề
N = 20

X_data = np.random.randn(N, 1) * 5
Y_data = polynomial(X_data)
# Ma trận đặc trưng [X^2, X, 1]
XX = np.hstack([X_data * X_data, X_data, np.ones_like(X_data)])

# Khởi tạo trọng số ngẫu nhiên w = [w1, w2, w3]
w = torch.randn(3, 1, requires_grad=True)
x_tensor = torch.tensor(XX, dtype=torch.float32)
y_tensor = torch.tensor(Y_data, dtype=torch.float32)
print("Hệ số ngẫu nhiên ban đầu w:\n", w.detach().numpy().flatten())
# Tối ưu hóa bằng NAdam
optimizer = torch.optim.NAdam([w], lr=0.02)
EPOCHS = 2000
for epoch in range(EPOCHS):
  optimizer.zero_grad()
  y_pred = x_tensor @ w
  mse = torch.mean(torch.square(y_tensor - y_pred))
  mse.backward()
  optimizer.step()
learned_w = w.detach().numpy().flatten()
print("\nKết quả sau 2000 EPOCHS")
print(f"w1 ≈ {learned_w[0]:.4f} (hệ số x^2, cần đạt: 2)")
print(f"w2 ≈ {learned_w[1]:.4f} (hệ số x,   cần đạt: -3)")
print(f"w3 ≈ {learned_w[2]:.4f} (hệ số tự do, cần đạt: 5)")
```
